```python
#Pattern matching with sequences
#Python 3.10 introduced structural pattern matching, which allows you to match patterns in data
#Example of pattern matching with sequences

def process_sequence(seq):
    match seq: #This is the subject that you want to match against
        case [x, y]:  # Matches a list with exactly two elements
            print(f'Got a list with two elements: {x} and {y}')
        case (x, y):  # Matches a tuple with exactly two elements
            print(f'Got a tuple with two elements: {x} and {y}')
        case _:  # Matches anything else
            print('Got something else')

process_sequence([1, 2])  # Output: Got a list with two elements: 1 and 2
process_sequence((3, 4))  # Output: Got a list with two elements: 3 and 4. This is very important to understand, it does not matter how you unpack the sequence, it will match as long as the structure is correct. So a tuple with two elements will match the case for a list with two elements, and vice versa.
process_sequence([1, 2, 3])  # Output: Got something else

#you can also hard code some patterns you expect in your subject
def process_sequence(seq):
    match seq:
        case [0, x]:  # Matches a list where the first element is 0
            print(f'Got a list starting with 0 and second element: {x}')
        case [x, 0]:  # Matches a list where the second element is 0
            print(f'Got a list with first element: {x} and second element 0')
        case _:  # Matches anything else
            print('Got something else')
    
process_sequence([0, 1])  # Output: Got a list starting with 0 and second element: 1
process_sequence([1, 0])  # Output: Got a list with first element: 1 and second element 0
process_sequence([1, 2])  # Output: Got something else


#Pattern matching with nested sequences

# Even though it looks like a switch case, it is more than that. There is a powerful concept of destructuring. Not only you unpack sequences into matching cases that share the same signatures, but you also can have conditions and gurads while you process that case.

def process_nested_sequence(seq):
    match seq:
        case [name, _, _, (lat, lon)] if lon < 0:  # Matches a list with a city name and a tuple of latitude and longitude`
        # "_" is a wildcard that matches any value but does not bind it to a variable. This is useful when you want to ignore certain elements in the sequence. It is the only symbol that can appear twice in the syntax.
        # You can also use "*_" to ignore multiple elements in a sequence.
            print(f'City: {name}, Latitude: {lat}, Longitude: {lon}')
        case _:  # Matches anything else
            print('Got something else')

process_nested_sequence(['New York', 'NY', 'USA', (40.7128, -74.0060)])  # Output: City: New York, Latitude: 40.7128, Longitude: -74.0060
process_nested_sequence(['Los Angeles', 'CA', 'USA', 34.0522, -118.2437])  # Output: Got something else. Because the structure of the case did not match. the structure expected lat and lon to be a tuple/sequence.-

#Note - The sequence pattern can match instances of most actual or virtual subclasses of collections.abc.Sequence, including lists, tuples. str, bytes, and bytesarray are exceptions.
```

```python
from collections import defaultdict

class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        response = defaultdict(list) # defualtdict is a subclass of dict that calls a factory function to supply missing values. In this case, the factory function is list, so if you try to access a key that does not exist in the dictionary, it will return an empty list instead of raising a KeyError.

        for s in strs:
            count = [0] * 26  # Create a list to count the occurrences of each letter (assuming only lowercase letters). This is how we can represent the frequency of each letter in the string. The index of the list corresponds to the letter (0 for 'a', 1 for 'b', etc.), and the value at that index represents how many times that letter appears in the string.
            for char in s:
                count[ord(char) - ord('a')] += 1  # Increment the count for the corresponding letter. ord(char) gives the ASCII value of the character, and ord('a') gives the ASCII value of 'a'. By subtracting ord('a') from ord(char), we get a zero-based index for the letter (e.g., 'a' -> 0, 'b' -> 1, ..., 'z' -> 25).
            response[tuple(count)].append(s)  # Use the count list as a key in the response dictionary. Since lists cannot be used as keys in a dictionary, we convert it to a tuple, which is immutable and can be used as a key. The value associated with this key is a list of strings that have the same letter frequency, i.e., they are anagrams of each other.
        return list(response.values())  # Return the grouped anagrams as a list of lists. response.values() gives us a view of the values in the dictionary, which are the lists of anagrams. We convert this view to a list before returning it.
```

```python
#slicing objects
# All sequence types in Python support slicing, which allows you to extract a portion of the sequence using a specific syntax. The slicing syntax is as follows: sequence[start:stop:step]. Here, start is the index where the slice begins (inclusive), stop is the index where the slice ends (exclusive), and step is the interval between elements in the slice.
my_list = [0, 1, 2, 3, 4, 5]
print(my_list[1:4])  # Output: [1, 2, 3]

#Why slicing ranges exlcude the last element
#The reason slicing ranges exclude the last element is because of how these sequence structures are written in C and how python itself manages indexes of sequences.

print(my_list[:3])  # Output: [0, 1, 2] if anything is missing in the start or stop, it is assumed to be the beginning or end of the sequence respectively. So in this case, it starts from the beginning of the list and goes up to index 3 (exclusive), which gives us the first three elements of the list.
print(my_list[3:])  # Output: [3, 4, 5] if the stop index is missing, it is assumed to be the beginning of the sequence. So in this case, it starts from index 3 and goes to the end of the list, which gives us the last three elements of the list.

s = "bicycle"
print(s[::3])  # Output: 'bye' This means we are slicing the string starting from the beginning to the end with a step of 3, which gives us every third character in the string.
print(s[::-1])  # Output: 'elcycib' This means we are slicing the string starting from the end to the beginning with a step of -1, which gives us the string in reverse order. This is a common idiom for reversing a sequence in Python.

#Slice objects are very powersful and can be used to represent the slicing operation in a more abstract way. They can be created using the slice() function, which takes the same arguments as the slicing syntax (start, stop, step).

#This is a literal string sequence that represents an invoice. The first line is a header that describes the columns of the invoice, and the subsequent lines represent individual items with their details such as item number, description, price, quantity, and total cost.
invoice = """
... 0.....6.................................40........52...55........
... 1909 Pimoroni PiBrella $17.50 3 $52.50
... 1489 6mm Tactile Switch x20 $4.95 2 $9.90
... 1510 Panavise Jr. - PV-201 $28.00 1 $28.00
... 1601 PiTFT Mini Kit 320x240 $34.95 1 $34.95
... """

#we can use the slice objects to extract specific parts of each line in the invoice. For example, we can create slice objects for the item number, description, price, quantity, and total cost columns based on the positions of the dots in the header line.
SKU = slice(0, 6)  # This slice object represents the item number column, which starts at index 0 and ends at index 6 (exclusive).
DESCRIPTION = slice(6, 40)  # This slice object represents the description column, which starts at index 6 and ends at index 40 (exclusive).
PRICE = slice(40, 52)  # This slice object represents the price column,which starts at index 40 and ends at index 52 (exclusive).
QUANTITY = slice(52, 55)  # This slice object represents the quantity column, which starts at index 52 and ends at index 55 (exclusive).
TOTAL = slice(55, None)  # This slice object represents the total cost column,  which starts at index 55 and goes to the end of the line (None indicates that it goes to the end).

for line in invoice.split('\n')[2:]:  # We split the invoice string into lines using the newline character as a delimiter, and then we skip the first two lines (the header and the empty line) by slicing the list of lines starting from index 2.
    print(line[SKU], line[DESCRIPTION], line[PRICE], line[QUANTITY], line[TOTAL])

#Multidimensional slicing
# The [] operator can also take multiple indexes or slices separated by commas, which allows you to perform multidimensional slicing on sequences that support it, such as lists of lists or NumPy arrays.
# Numpy arrays are 2D arrays so you can somewhat sort them using notation like arr[1:3, 2:4] which will give you a 2D array with the elements from rows 1 to 2 and columns 2 to 3. This is a powerful way to extract specific subarrays from a larger array.

#Assigning to slices
#Mutable sequences can be modified in place using the slice notations on the left hand side of an assignment. This allows you to replace a portion of the sequence with new values, which can be very useful for tasks like inserting, deleting, or replacing elements in a list.

my_list = [0, 1, 2, 3, 4, 5]
my_list[2:5] = [20, 30]  # This replaces the elements at index 2, 3, and 4 with the new values 20 and 30. The element at index 2 is replaced with 20, and the elements at index 3 and 4 are replaced with 30. Since we are replacing three elements with two new values, the list will be modified to have one less element.
print(my_list)  # Output: [0, 1, 20, 30
del my_list[2:4]  # This deletes the elements at index 2 and 3 from the list. After this operation, the list will have two fewer elements.
print(my_list)  # Output: [0, 1, 5]
my_list[1:3] = 100 #this will be an error because the right side has to be an iterable, so we need to wrap 100 in a list or a tuple to make it an iterable with one element.
```

```python
#Using + and * with sequences

# + and * operators can be used with sequences to perform concatenation and repetition, respectively. They always return a new sequence and do not modify the original sequence in place.

list1 = [1, 2, 3]
list2 = [4, 5, 6]
concatenated_list = list1 + list2  # This creates a new list that is the concatenation of list1 and list2.
print(concatenated_list)  # Output: [1, 2, 3
print(list1)  # Output: [1, 2, 3] The original list1 remains unchanged.
my_list = [0, 1, 2]
repeated_list = my_list * 3  # This creates a new list that is the repetition of my_list three times.
print(repeated_list)  # Output: [0, 1, 2, 0, 1, 2, 0, 1, 2]
print(my_list)  # Output: [0, 1, 2] The original my_list remains unchanged.


#Building Lists of list and behavior of + and * with nested lists
#When you use the + operator with nested lists, it concatenates the outer lists but does not create new inner lists. This means that the inner lists are shared between the original lists and the concatenated list. As a result, if you modify an inner list in one of the original lists, it will also affect the concatenated list because they reference the same inner list.
list1 = [[1, 2], [3, 4]]
list2 = [[5, 6], [7, 8]]
concatenated_list = list1 + list2  # This concatenates the outer lists, resulting in a new list that contains the inner lists from both list1 and list2.
print(concatenated_list)  # Output: [[1, 2], [3, 4], [5, 6], [7, 8]]
list1[0][0] = 'X'  # This modifies the first element of the first inner list in list1.
print(concatenated_list)  # Output: [['X', 2], [3, 4], [5, 6], [7, 8]] The modification to list1 also affects concatenated_list because they share the same inner lists.
```

```python
#list.sort vs sorted
#The list.sort() method sorts a list in place, meaning that it modifies the original list and does not return a new list. On the other hand, the sorted() function returns a new sorted list without modifying the original list. This is an important distinction to understand when working with sorting in Python, as it affects how you can use these functions and whether you need to keep a reference to the original list.
# In python, as a rule the methods that modify an object in place return None, while functions that return a new object return the new object. This is a convention in Python that helps to make it clear whether an operation modifies an object or returns a new one.
my_list = [3, 1, 2]
my_list.sort()  # This sorts the list in place.
print(my_list)  # Output: [1, 2, 3]

my_list = [3, 1, 2]
sorted_list = sorted(my_list)  # This returns a new sorted list.
print(sorted_list)  # Output: [1, 2, 3]
print(my_list)  # Output: [3, 1, 2] The original list remains unchanged.

#both these functions can take a key function and a reverse flag to customize the sorting behavior. The key function is a function that takes an element as input and returns a value that will be used for sorting. The reverse flag is a boolean that indicates whether the sorting should be done in reverse order (descending) instead of the default ascending order.
my_list = ['apple', 'banana', 'cherry']
my_list.sort(key=len)  # Sort by length of each string.
print(my_list)  # Output: ['apple', 'banana', 'cherry']

my_list = ['apple', 'banana', 'cherry']
sorted_list = sorted(my_list, key=len, reverse=True)  # Sort by length in descending order.

print(sorted_list)  # Output: ['banana', 'apple', 'cherry']
print(my_list)  # Output: ['apple', 'banana', 'cherry'] The original list remains unchanged.
```

```python
#deque and other queues

#.append or .pop methods make a list usable as stack and .append and .pop(0) make it usable as a FIFO queue. However, using a list as a queue can be inefficient because .pop(0) has to shift all the other elements in the list, which can lead to O(n) time complexity for each pop operation. 
# In contrast, a deque (double-ended queue) from the collections module is optimized for fast appends and pops from both ends, making it more efficient for use as a queue. 
# A deque allows you to append and pop elements from both ends in O(1) time complexity, which can significantly improve performance when working with queues compared to using a list.
# A perfect example for implementation can be "last seen items" functionality. A dequeu can be bounded to a certain size, and when new items are added beyond that size, the oldest items are automatically removed from the deque. This makes it an ideal data structure for keeping track of a fixed number of recent items, such as the last seen items in a web application or the most recent messages in a chat application.

from collections import deque
my_deque = deque([1, 2, 3, 4, 5])
print(sum(my_deque))  # This calculates the sum of all the elements in the deque, which is 15 in this case.
print(my_deque)  # Output: deque([1, 2, 3, 4, 5]) This is the deque that was created from the list. It is a double-ended queue that contains the elements 1, 2, 3, 4, and 5.
my_deque.append(6)  # This appends the value 6 to the right end of the deque.
print(my_deque)  # Output: deque([1, 2, 3, 4, 5, 6]) This shows that the value 6 has been added to the right end of the deque.
my_deque.appendleft(0)  # This appends the value 0 to the left end of the deque.
print(my_deque)  # Output: deque([0, 1, 2, 3, 4, 5, 6]) This shows that the value 0 has been added to the left end of the deque.
my_deque.pop()  # This removes and returns the value from the right end of the deque, which is 6 in this case.
print(my_deque)  # Output: deque([0, 1, 2, 3, 4, 5]) This shows that the value 6 has been removed from the right end of the deque.
my_deque.popleft()  # This removes and returns the value from the left end of the deque, which is 0 in this case.
print(my_deque)  # Output: deque([1, 2, 3, 4, 5]) This shows that the value 0 has been removed from the left end of the deque.

#Notice all this is happening inplace, the original deque is being modified directly without creating a new object in memory, which is one of the key advantages of using a deque for queue-like operations.

my_deque.rotate(3)  # This rotates the deque to the right by 3 steps. The last 3 elements will be moved to the front of the deque.
print(my_deque)  # Output: deque([3, 4, 5, 1, 2]) This shows that the deque has been rotated to the right by 3 steps, resulting in the last 3 elements (3, 4, 5) being moved to the front of the deque.
my_deque.rotate(-2)  # This rotates the deque to the left by 2 steps. The first 2 elements will be moved to the end of the deque.
print(my_deque)  # Output: deque([5, 1, 2, 3, 4]) This shows that the deque has been rotated to the left by 2 steps, resulting in the first 2 elements (3, 4) being moved to the end of the deque.
```

```python
import math

math.lcm(2,3,4)

nums = [1,2,3,4]

#The most optmized solution is to calculate the product of all the elements in the list and then divide that product by each element to get the result for each position. This way we only need to calculate the product once and then perform n divisions, which is more efficient than calculating the product for each position separately.
# However, we need to be careful with this approach if there are zeros in the list, as it will lead to division by zero. In that case, we can handle the zeros separately by counting how many zeros are in the list and adjusting the product accordingly.
# So first check the number of zeroes, if it is more than 1,then result will be all zeroes. If it is exactly 1, then the result will be zero for all positions except the position of the zero, which will be the product of all non-zero elements. If there are no zeroes, then we can proceed with the original approach of calculating the product and dividing by each element.

#However, the problem can also be solved without using division, it is done by calculating prefix and suffix product arrays. i.e. for an index calculate the products of all prefixed elements i.e. on the left side and the products of all suffixed elements i.e. on the right side, and then multiply these two products to get the result for that index. This way we can avoid division and also handle the case of zeros more gracefully.

def productExceptSelf(nums):
    n = len(nums)
    ans = [1] * n  # Initialize the answer array with 1s
    current_product = 1
    # Calculate prefix products
    for i in range(n):
        ans[i] *= current_product  # Store the product of all elements to the left of index i in the answer array
        current_product *= nums[i]  # Update the current product for the next iteration
    
    current_product = 1
    # Calculate suffix products, range takes n-1, i.e. the second last index, down to -1, remember range does not include the end value and step is -1, i.e. we are moving backwards through the array
    for i in range(n - 1, -1, -1): 
        ans[i] *= current_product  # Multiply the product of all elements to the right of index i with the existing value in the answer array
        current_product *= nums[i]  # Update the current product for the next iteration
    
    return ans


productExceptSelf(nums)
```

```python
board = [["5","3",".",".","7",".",".",".","."],["6",".",".","1","9","5",".",".","."],[".","9","8",".",".",".",".","6","."],["8",".",".",".","6",".",".",".","3"],["4",".",".","8",".","3",".",".","1"],["7",".",".",".","2",".",".",".","6"],[".","6",".",".",".",".","2","8","."],[".",".",".","4","1","9",".",".","5"],[".",".",".",".","8",".",".","7","9"]]

print(board)
print(f"\n")
zip(*board) #This is a common technique to transpose a matrix in Python. The zip(*board) expression takes the rows of the board and zips them together, effectively creating columns from the rows. The * operator is used to unpack the rows of the board as separate arguments to the zip function. The result is an iterator of tuples, where each tuple contains the elements of a column from the original board. By converting this iterator to a list
from typing import List
def isValidSudoku(board: List[List[str]]) -> bool:

    def is_valid_group(group):
        seen = set()
        for cell in group:
            if cell != ".":
                if cell in seen:
                    return False
                seen.add(cell)
        return True

    # Check rows
    for row in board:
        if not is_valid_group(row):
            return False
    
    # Check columns
    for col in zip(*board):  # Transpose the board to get columns as rows
        if not is_valid_group(col):
            return False
    
    # Check 3x3 sub-boxes
    for i in range(0, 9, 3):
        for j in range(0, 9, 3):
            box = [board[x][y] for x in range(i, i + 3) for y in range(j, j + 3)]
            if not is_valid_group(box):
                return False
    
    return True

#The above example O(n^2) time complexity because we are checking each row, column, and box which takes O(n) time and we are doing this for n rows, n columns, and n/3 boxes (since each box is 3x3), resulting in O(n^2) overall. The space complexity is O(n) due to the set used to track seen numbers in each group.

#If we want to solve this problem in O(n) time complexity, we can use a single pass approach where we check the validity of the board while iterating through it only once. 
# We can maintain three sets for each row, column, and box to track the numbers we have seen so far. As we iterate through each cell, we can check if the number has already been seen in the corresponding row, column, or box. If it has, then the board is not valid. If it hasn't, we add it to the respective sets and continue checking until we have processed all cells.

def isValidSudoku(board: List[List[str]]) -> bool:
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = [set() for _ in range(9)]

    for i in range(9):
        for j in range(9):
            cell = board[i][j]
            if cell == ".":
                continue

            box_index = (i // 3) * 3 + (j // 3) # Calculate the index of the 3x3 box based on the current row and column indices. The formula (i // 3) * 3 + (j // 3) divides the board into 3x3 boxes and assigns a unique index to each box. For example, the top-left box will have an index of 0, the top-middle box will have an index of 1, and so on.

            if cell in rows[i] or cell in cols[j] or cell in boxes[box_index]:
                return False

            rows[i].add(cell)
            cols[j].add(cell)
            boxes[box_index].add(cell)

    return True
```

```python
#Automatic handling of missing keys with defaultdict
# The defaultdict class from the collections module is a subclass of the built-in dict class that provides a default value for missing keys.
from collections import defaultdict
my_defaultdict = defaultdict(list)  # This creates a defaultdict where the default value for any missing key will be an empty list.
my_defaultdict['a'].append(1)  # This will automatically create a new entry for the key 'a' with an empty list as the default value, and then append the value 1 to that list.
print(my_defaultdict)  # Output: defaultdict(<class 'list'>, {'a': [1]}) This shows that the key 'a' has been added to the defaultdict with a list containing the value 1. The output also indicates that this is a defaultdict with a default factory of type list.

#possible use cases -
# 1. Grouping items: You can use a defaultdict to group items based on a common key. For example, you can group a list of words by their first letter using a defaultdict with a list as the default factory.
words = ['apple', 'banana', 'cherry', 'avocado', 'blueberry']
grouped_words = defaultdict(list)
for word in words:
    grouped_words[word[0]].append(word)
print(grouped_words)  # Output: defaultdict(<class 'list'>, {'a': ['apple', 'avocado'], 'b': ['banana', 'blueberry'], 'c': ['cherry']})

# 2. Counting occurrences: You can use a defaultdict to count the occurrences of items in a collection. For example, you can count the frequency of characters in a string using a defaultdict with an integer as the default factory.
text = "hello world"
char_count = defaultdict(int)

for char in text:
    char_count[char] += 1
print(char_count)  # Output: defaultdict(<class 'int'>, {'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 1, 'w': 1, 'r': 1, 'd': 1})
```

```python
height = [1,8,6,2,5,4,8,3,7]

def maxArea(self, height: List[int]) -> int:
        left = 0
        right = len(height) - 1
        max_area = 0

        while(left < right):
            if height[left] > height[right]:
                area = height[right] * (right - left)
                right -= 1
            else:
                area = height[left] * (right - left)
                left += 1
            max_area = max(max_area, area)

        return max_area
        
        
        
        


maxArea(height)
```

```python
#Inconsistent using of __missing__ in standard library
# The __missing__ method is a special method that can be defined in a subclass of dict to provide a default value for missing keys. 
# Or it can be used to raise exceptions or perform other custom behavior when a key is not found in the dictionary.
# However, it is not just about defining the __missing__ method. Different methods in base classes sometimes call __missing__ and sometimes they don't, which can lead to inconsistent behavior when using the dictionary.

class MyDict(dict): #This is a custom dictionary class that inherits from the built-in dict class. By defining the __missing__ method, we can specify what should happen when a key is accessed that does not exist in the dictionary.
    def __missing__(self, key):
        return f"Key {key} is missing"
    
my_dict = MyDict({'a': 1, 'b': 2})
print(my_dict['a'])  # Output: 1 This shows that the existing key 'a' returns its corresponding value, which is 1.
print(my_dict['c'])  # Output: Key c is missing This shows that when we try to access the key 'c', which does not exist in the dictionary, the __missing__ method is called, and it returns the string "Key c is missing". This demonstrates how the __missing__ method provides a default response for missing keys in the custom dictionary class.

#Not all mapping types behave the same way with __missing__. For example, the built-in dict class does not call __missing__ when a key is accessed that does not exist, and instead raises a KeyError. On the other hand, the collections.defaultdict class does call __missing__ to provide a default value for missing keys. This inconsistency can lead to confusion when working with different mapping types, as the behavior for missing keys may not be uniform across all mappings.
```

```python
#Variations of dict

# defaultdict: A subclass of dict that provides a default value for missing keys. 
# It allows you to specify a default factory function that will be called to provide a default value when a key is accessed that does not exist in the dictionary.

#collections.OrderedDict: A subclass of dict that maintains the order of keys as they were inserted. After python 3.7, the built-in dict also maintains insertion order, so the need for OrderedDict has diminished in many cases. 
# However, OrderedDict still provides some additional methods that are not available in the built-in dict, such as move_to_end() and popitem(last=False), which can be useful for certain applications.
# The equality operation for OrderedDict considers the order of keys, meaning that two OrderedDict instances are considered equal only if they have the same key-value pairs in the same order. In contrast, the equality operation for the built-in dict does not consider the order of keys, so two dict instances are considered equal if they have the same key-value pairs regardless of their order.

# collections.ChainMap: A class that groups multiple dictionaries together to create a single, updateable view. 
# It allows you to search through multiple dictionaries as if they were a single dictionary, which can be useful for managing contexts or configurations.
# The lookup happens for each mapping in the sequence they were provided to the chainmap constructor. So if you have multiple dictionaries in a ChainMap, the lookup will first check the first dictionary, then the second, and so on, until it finds the key or exhausts all dictionaries.

MyDict = {'a': 1, 'b': 2}
MyDict2 = {'b': 3, 'c': 4}
from collections import ChainMap
chain_map = ChainMap(MyDict, MyDict2)
print(chain_map['a'])  # Output: 1 This shows that the key 'a' is found in the first dictionary (MyDict) of the ChainMap, and its value is returned.
print(chain_map['b'])  # Output: 2 This shows that the key 'b' is found in the first dictionary (MyDict) of the ChainMap, and its value is returned. Even though 'b' also exists in the second dictionary (MyDict2), the value from the first dictionary takes precedence in the lookup.
print(chain_map['c'])  # Output: 4 This shows that the key 'c' is not found in the first dictionary (MyDict) of the ChainMap, so the lookup continues to the second dictionary (MyDict2), where it is found, and its value is returned.

#chainmap does not copy the input mappings, it creates a view that references the original dictionaries. This means that if you modify one of the original dictionaries, the changes will be reflected in the ChainMap. For example, if you update MyDict['a'] = 10, then chain_map['a'] will also return 10, demonstrating that the ChainMap is referencing the original dictionary and not creating a separate copy.
#Updates the chainmap will only affect the first dictionary in the chain. For example, if you do chain_map['a'] = 10, it will update MyDict['a'] to 10, but it will not affect MyDict2. This is because the ChainMap is designed to allow updates to the first dictionary while still providing a view of all the dictionaries in the chain for lookups.

#collections.Counter: A subclass of dict that is used to count hashable objects. It provides a convenient way to count the occurrences of items in a collection, and it includes methods for common operations like finding the most common elements or subtracting counts.
from collections import Counter
ct = Counter('abracadabra')  # This creates a Counter object that counts the occurrences of each character in the string 'abracadabra'.
print(ct)  # Output: Counter({'a': 5, 'b': 2, 'r': 2, 'c': 1, 'd': 1}) This shows the count of each character in the string, with 'a' appearing 5 times, 'b' and 'r' appearing 2 times each, and 'c' and 'd' appearing once each.
print(ct.most_common(2))  # Output: [('a', 5), ('b', 2)] This shows the two most common characters in the Counter, which are 'a' with a count of 5 and 'b' with a count of 2. The most_common() method returns a list of tuples, where each tuple contains an element and its corresponding count, sorted in descending order of count.


#shelve.Shelf: provides persistent storage for a mapping of string keys to arbitrary Python objects, like pickled objects. 
# The curious name comes from the fact that it is a "shelf" for storing objects, and it is built on top of the pickle module to serialize and deserialize objects for storage. 
# It allows you to store and retrieve Python objects in a file-based database, making it useful for simple data persistence without the need for a full-fledged database system.

#shelve.open function is used to create a new shelf or open an existing one. It takes the filename as an argument and returns a Shelf object that can be used to store and retrieve data. The data is stored in a file on disk, and the Shelf object provides a dictionary-like interface for accessing the data. You can use it like a regular dictionary, but the data will be persisted across program runs, allowing you to save and load data easily.
```

```python
nums = set([100,4,200,1,3,2])

def longestConsecutive(nums):
    longest_streak = 0
    num_set = set(nums)  # This creates a set from the list of numbers, which allows for O(1) average time complexity for lookups.

    for num in num_set:  # This iterates through each number in the set.
        if num - 1 not in num_set:  # This checks if the current number is the start of a sequence by checking if the previous number (num - 1) is not in the set.
            current_num = num  # If it is the start of a sequence, we initialize current_num to the current number.
            current_streak = 1  # We also initialize current_streak to 1, since we have at least one number in the sequence.

            while current_num + 1 in num_set:  # This loop continues to check for the next number in the sequence (current_num + 1) and increments the current_streak count as long as the next number is found in the set.
                current_num += 1
                current_streak += 1

            longest_streak = max(longest_streak, current_streak)  # After exiting the inner loop, we update longest_streak with the maximum of its current value and the current_streak.

    return longest_streak  # Finally, we return the longest streak found.

print(longestConsecutive(nums))  # Output: 4 This shows that the longest consecutive sequence in the input list is of length 4, which corresponds to the sequence [1, 2, 3, 4].
```

```python
nums = [3, 6, 8, 10, 1, 2, 1]

def partition(lst: list[int], low:int, high:int):
            pivot = lst[high]
            index = low - 1

            for i in range(low, high):
                if lst[i] <= pivot:
                    index += 1
                    lst[i], lst[index] = lst[index], lst[i]

            lst[index+1], lst[high] = lst[high], lst[index+1]
            return index + 1

def qs(lst: list[int], low:int, high:int):
    if(low<high):
        part_index = partition(lst, low, high)
        qs(lst, low, part_index - 1)
        qs(lst, part_index + 1, high)
        
def quickSort(lst: list[int]):
    qs(lst, 0, len(lst) - 1)


quickSort(nums)

print(nums)
```

```python
from collections import deque
from typing import List
class Solution:
    def trap(self, height: List[int]) -> int:
        result = 0
        n = len(height)

        for i in range(n):
            if i == n-1:
                break
                
            depression = deque()
            if height[i+1] < height[i] and i+1 != n-1:
                trap = True
                depression.append(height[i])
                bound_start = i
                while trap and i < n-2:
                    i += 1
                    if height[i] >= height[bound_start]:
                        depression.append(height[i])
                        trap = False
                    else:
                        depression.append(height[i])
                
                water_level = min(depression.pop(), depression.popleft())
                max_water = water_level * (i-bound_start)
                result += max_water - sum(depression)
        
        return result

                


height = [0,1,0,2,1,0,1,3,2,1,2,1]

s = Solution()
print(s.trap(height))  # Output: 6 This shows that the total amount of trapped rainwater between the bars represented by the input list is 6 units.
```

```python
def repeatingCharacters(s: str, k: int) -> int:
    char_count = defaultdict(int)
    left = 0
    max_length = 0

    def isValidSubstring(input: dict) -> bool:
        sorted_values = sorted(input.values())
        sorted_values.remove(max(sorted_values))
        return sum(sorted_values) <= k
    

    for right in range(len(s)):
        char_count[s[right]] += 1

        while not isValidSubstring(char_count):
            char_count[s[left]] -= 1
            if char_count[s[left]] == 0:
                del char_count[s[left]]
            left += 1
        
        max_length = max(max_length, right - left + 1)

    return max_length
```
