```python
#Static Array

#Read

arr = [1, 2, 3, 4, 5]

print(arr[2])  # Output: 3


#traversing through the array

for i in range(len(arr)):
    print(arr[i])  # Output: 1 2 3 4 5

#OR

while i < len(arr):
    print(arr[i])  # Output: 1 2 3 4 5
    i += 1

#Deleting from array, in static array you cannot delete from memory, you can only delete or replace the value at a specific index by another value let's say -1

#deleting from end

if len(arr) > 0:
    arr[len(arr) - 1] = -1  # Mark the last element as deleted

#deleting ith index

def delete_at_index(arr, index):

    for i in range(index + 1, len(arr)):
        arr[i - 1] = arr[i]  # Shift elements to the left


#Insert at end

def insert_at_end(arr, value, capacity):
    if len(arr) < capacity:
        arr[len(arr)] = value  # Insert the value at the end
    else:
        print("Array is full. Cannot insert.")


#Insert at ith index

def insert_at_index(arr, index, value, capacity):
    if len(arr) < capacity:
        for i in range(len(arr), index, -1):
            arr[i] = arr[i-1]  # Shift elements to the right
        arr[index] = value  # Insert the new value at the specified index
    else:
        print("Array is full. Cannot insert.")
```

```python
#Max consecutive numbers

#You are given a binary array nums, return the maximum number of consecutive 1's in the array.

# Example 1:

# Input: nums = [1,1,0,1,1,1]

# Output: 3

# Example 2:

# Input: nums = [1,0,1,1,0,1]

# Output: 2


from typing import List


class Solution:
    def findMaxConsecutiveOnes(self, nums: List[int]) -> int:
        max_count = 0

        count = 0

        for num in nums:
            if num == 1:
                count += 1
                max_count = max(max_count, count)
            else:
                count = 0

        return max_count


#call
solution = Solution()
result = solution.findMaxConsecutiveOnes([1,1,0,1,1,1])
print(result)  # Output: 3
```

```python
# Given an integer array nums and an integer val, remove all occurrences of val in nums in-place. The order of the elements may be changed. Then return the number of elements in nums which are not equal to val.

# Consider the number of elements in nums which are not equal to val be k, to get accepted, you need to do the following things:

# Change the array nums such that the first k elements of nums contain the elements which are not equal to val. The remaining elements of nums are not important as well as the size of nums.
# Return k.
# Custom Judge:

# The judge will test your solution with the following code:

# int[] nums = [...]; // Input array
# int val = ...; // Value to remove
# int[] expectedNums = [...]; // The expected answer with correct length.
#                             // It is sorted with no values equaling val.

# int k = removeElement(nums, val); // Calls your implementation

# assert k == expectedNums.length;
# sort(nums, 0, k); // Sort the first k elements of nums
# for (int i = 0; i < k; i++) {
#     assert nums[i] == expectedNums[i];
# }

# Example 1:

# Input: nums = [3,2,2,3], val = 3

# Output: k = 2, nums = [2,2,_,_]
# Explanation: Your function should return k = 2, with the first two elements of nums being 2.
# It does not matter what you leave beyond the returned k (hence they are underscores).


# Example 2:

# Input: nums = [0,1,2,2,3,0,4,2], val = 2

# Output: k = 5, nums = [0,1,3,0,4,_,_,_]
# Explanation: Your function should return k = 5, with the first five elements of nums containing 0, 0, 1, 3, and 4.
# Note that the five elements can be returned in any order.
# It does not matter what you leave beyond the returned k (hence they are underscores).


class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:

        k=0 #write pointer for the next position to write a non-val element

        for num in nums:
            if num != val:
                nums[k] = num
                #k only moves forward when we find a non-val element to write, otherwise it stays in place and we overwrite the val elements
                k += 1

        return k

#efficient implementation - we move the val element to the end of the array and reduce the size of array by 1 each time we find a val element, this way we don't have to shift elements and we can do it in O(n) time complexity and O(1) space complexity
# we greatly reduce the number of writes to the array, which can be beneficial if the array is large and the number of val elements is small.

    def removeElementEfficient(self, nums: List[int], val: int) -> int:

        n = len(nums)
        i = 0

        while i < n:
            if nums[i] == val:
                n -= 1 #to reduce the size of the array by 1 early to avoid OOM
                nums[i] = nums[n] #move the last element to the current position
            else:
                i += 1 #only move forward if we didn't find a val element, otherwise we need to check the new element at the current position

        return n

#call
solution = Solution()
result = solution.removeElement([3,2,2,3], 3)
print(result)  # Output: 2, nums = [2,2,_,_]

result = solution.removeElementEfficient([0,1,2,2,3,0,4,2], 2)
print(result)  # Output: 5, nums = [0,1,3,0,4,_,_,_]
```

```python
#Dynamic Array

# As the name suggests, a dynamic array is an array that can grow and shrink in size. It is implemented using a static array, but it automatically resizes itself when it reaches its capacity.

#add to last

def add_to_last(self, arr, value):
    if self.length == self.capacity:
        self.resize(arr)  # Resize the array if it's full

    arr[self.length] = value  # Add the value at the end
    self.length += 1  # Increment the length of the array

def resize(self, arr):
    new_capacity = self.capacity * 2  # Double the capacity
    new_arr = [0] * new_capacity  # Create a new array with the new capacity

    for i in range(self.length):
        new_arr[i] = arr[i]  # Copy elements to the new array

    self.capacity = new_capacity  # Update the capacity
    self.arr = new_arr  # Update the reference to the new array

def remove_from_last(self, arr):
    if self.length == 0:
        print("Array is empty. Cannot remove.")
        return

    arr[self.length - 1] = 0  # Mark the last element as removed
    self.length -= 1  # Decrement the length of the array

def get(self, arr, index):
    if index < 0 or index >= self.length:
        print("Index out of bounds.")
        return None

    return arr[index]  # Return the value at the specified index

def set(self, arr, index, value):
    if index < 0 or index >= self.length:
        print("Index out of bounds.")
        return

    arr[index] = value  # Set the value at the specified index
```
