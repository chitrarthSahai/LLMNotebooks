```python
#Regular Expressions are a notation to define a pattern of text we need to search for. Read from left to right.

#search in re
import re

text = "This is a sample text"
pattern = "sample"

match =  re.search(pattern, text)

print("Found a match") if match else print("Match not found")
```

```python
#re match

pattern = r'The'
text = 'The quick brown fox'

match = re.match(pattern, text)

print("Found a match") if match else print("Match not found")
```

```python
#re fullmatch -  checks if full string matches regex

pattern = r'\d{4}-\d{2}-\d{2}'
text = '2023-10-05'

match = re.fullmatch(pattern, text)

print("Found a match") if match else print("Match not found")
```

```python
#re findall - searches a string for all non-overlapping matches

pattern = r'\d+'
text = 'Thr price of the book is $19.99 and the price of the movie is $12.50'

match = re.findall(pattern, text)

print("Found a match") if match else print("Match not found")
print(match)
```

```python
#Sub and Subn function - Search a string for all non overlapping occurrences and the replace them with a string.

#Accepts five parameters: pattern, string to be searched, replacement, optional max number of replacements, and optional flags.
#Sub does not return anything, but subn returns a new string and number of replacements made.

pattern = r'\b\w{4}\b' #/b are word boundaries and \w{4} means a word with exactly 4 letters.
text = 'The quick brown fox jumped over a lazy dog.'

new_text = re.sub(pattern, '****', text)

print(new_text)
```

```python
#Split function - Really helpful. It splits a string into optional number of max splits based on regex match occurence.

pattern = r'\s+'
text = 'The quick brown fix jumps over the lazy dog'

words = re.split(pattern, text)

print(words)
```

```python
#Compile function - A very useful function to compile a pattern into a regex object that can be passed to several functions and classes and a pattern can be reused for multiple scenarios.

pattern = r'\d+'

text = 'The price of the book is $19.99 and the price of the movie is $12.50'

regex = re.compile(pattern)
matches = regex.findall(text)

print(matches)
```

```python
#Validating user input and data
import re

#Often used to validate and perform sanity checks for data validation scenarios. For example, validating email addresses, phone numbers, postal codes, etc. before storing them in a database.

def validate_username(username):
    pattern = r'^[a-zA-Z0-9_]{3,20}$'  #Username must be 3-20 characters long and can contain letters, numbers, and underscores
    match = re.fullmatch(pattern, username)

    return match is not None

user_input = input("Enter a username: ")

if validate_username(user_input):
    print(f"{user_input} is a valid username")
else:
    print("Invalid username")


#Checking numeric data

def validate_strictly_numeric(input_string):
    pattern = r'^\d+$'  #String must contain only digits from start to end
    match = re.fullmatch(pattern, input_string)

    return match is not None

user_input = input("Enter a numeric string: ")

if validate_strictly_numeric(user_input):
    print(f"{user_input} is a valid numeric string")
else:
    print("Invalid numeric string")

#checking if valid email address

def validate_email(email):
    pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}$'  #Basic email validation pattern
    match = re.fullmatch(pattern, email)

    return match is not None

user_input = input("Enter an email address: ")

if validate_email(user_input):
    print(f"{user_input} is a valid email address")
else:
    print("Invalid email address")


#Sanitize input as measure against injection attacks - Code injection, SQL injection, is a dangerous security vulnerability.

def sanitize_input(user_input):
    pattern = r'[;\'"--]'  #Pattern to match potentially dangerous characters
    sanitized_input = re.sub(pattern, '', user_input)  #Remove dangerous characters
    return sanitized_input

user_input = input("Enter some input: ")
safe_input = sanitize_input(user_input)
print("Original input:", user_input)
print(f"Sanitized input: {safe_input}")
```

```python
#Assertions for preceding and following the pattern

#Lookahead assertion - Asserts that a given pattern is followed by another pattern without including it in the match.
pattern = r'\b\w+(?=\sis\b)'  #Matches words followed by 'is' without including 'is' in the match
text = 'This is a test. That is another test.'
matches = re.findall(pattern, text)
print(matches)  #Output: ['This', 'That']

#Lookbehind assertion - Asserts that a given pattern is preceded by another pattern without including it in the match.
pattern = r'(?<=\b[Tt]he\s)\w+\b'  #Matches words preceded by 'the' without including 'the' in the match
text = 'The cat sat on the mat.'
matches = re.findall(pattern, text)
print(matches)  #Output: ['cat', 'mat']
```
